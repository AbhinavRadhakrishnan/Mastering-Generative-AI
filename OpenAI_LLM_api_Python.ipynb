Chat Completion

```python
from openai import OpenAI

client = OpenAI()


completion = client.chat.completions.create(
    model = "gpt-3.5-turbo-0125",
    messages=[
        {"role":"system","content": "you are a helpful assistant"},
         {"role":"user","content": "What is ML?"}

    ]
)
print(completion.choices[0].message.content)
```

```python
from openai import OpenAI

client = OpenAI()
#maintain chat history in a list
chat_history=[
        {"role":"system","content": "you are a helpful assistant"},
         {"role":"user","content": "What is ML?"},
        {"role":"assistant","content": "ML is a field of AI"},
         {"role":"user","content": "Give example"}


    ]
  #send the full history to api
completion = client.chat.completions.create(
    model = "gpt-3.5-turbo-0125",
    messages=chat_history
)

response = completion.choices[0].message.content
print(response)

#add the assistant's reply back into history for next round
chat_history.append({"role":"assistant","content":"responce"})
```
