```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser #raw output from llm to string output
from langchain_ollama import ChatOllama
```

```python
llm = ChatOllama(
    model = "gemma2:2b",
    temperature = 0.1,
)
```

```python
parser = StrOutputParser()
```

```python
def chat():
    chat_history = [
        ("system","you are a helpful chatbot.Be concise and accurate.")

    ]

    print("Langchain Chatbot. Type 'exit' to quit \n ")

    while True:
        user_input = input("User: ").strip()

        if user_input.lower() == "exit":
            break

        #add user message

        chat_history.append(("user",user_input)) #appending a tuple to view as individual


        # build prompts dynamically

        prompt = ChatPromptTemplate.from_messages(chat_history) #prebuild template is present in Langchain

        
        chain = prompt | llm | parser

        #get response for user query

        response = chain.invoke({})

        #show And save the assistant response

        print(f"Bot : {response}\n")

        chat_history.append(("assistant",response))

        print("-"*80)

        
     
```

```python
chat()
```
