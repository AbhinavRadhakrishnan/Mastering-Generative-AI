```python
from google import genai
client = genai.Client()

response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = "explain how AI works in few words"
)
print(response.text)
```

How to do chat request / multi-churn conversation

```python
chat = client.chats.create(model = "gemini-2.5-flash")
response = chat.send_message("AI is the future world")
print(response.text)

print("."*50)

response = chat.send_message("how can you say that")
print(response.text)
#by using chat LLM knows what we asked about previously
```
