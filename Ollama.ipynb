```python
import ollama

response = ollama.chat(
    model = "gemma2:2b",
    messages=[
        {"role":"system","content":"You are a knowledgable assistant "},
        {"role":"user","content":"Explain as simple as possible what is ml in few lines"},

    ]
)
print(response["message"]["content"])
```

in the above code we have to wait till the model generates the full code but in the code below we get the answers like stream format

```python
import ollama

stream = ollama.chat(
    model = "gemma2:2b",
    messages=[
        {"role":"system","content":"You are a knowledgable assistant "},
        {"role":"user","content":"Explain as simple as possible what is GenAI in few lines"},

    ],
    stream = True
)

for chunk in stream:
    print(chunk["message"]["content"],end="",flush = True)
```
